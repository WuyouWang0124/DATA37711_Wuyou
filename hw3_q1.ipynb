## (a) Poisson model and exponential link

### (i) Why is Poisson a reasonable choice?

The Poisson distribution is a reasonable choice because $Y$ is count data, taking non-negative integer values. It models these outcomes using a positive rate parameter.

### (ii) Why use the exponential function?

Each regression tree outputs real numbers, so $f(x)$ can be any real number. The exponential function ensures that the Poisson rate is positive:

$$
\mu(x)=\exp(f(x))>0.
$$

Under this model,

$$
E[Y\mid X=x]=\exp(f(x)).
$$


## (b) Loss function

The conditional probability mass function is

$$
P(Y=y\mid X=x)=\frac{\exp(-\exp(f(x)))\exp(yf(x))}{y!}.
$$

Taking the negative logarithm gives

$$
L(y,f(x))=\exp(f(x))-yf(x)+\log(y!).
$$

Since $\log(y!)$ does not depend on $f$, we can use

$$
\boxed{L(y,f(x))=\exp(f(x))-yf(x).}
$$

For $n$ training observations, the empirical loss is

$$
\frac{1}{n}\sum_{i=1}^{n}\left[\exp(f(x_i))-y_i f(x_i)\right].
$$


## (c) Derivative

Differentiating with respect to $f(x)$,

$$
\boxed{\frac{\partial L(y,f(x))}{\partial f(x)}=\exp(f(x))-y.}
$$

Therefore, the negative gradient is

$$
y-\exp(f(x)).
$$


## (d) Choosing each regression tree

At iteration $m$, let the current predictor be $f_{m-1}$.

First, calculate the negative gradient for each training observation:

$$
r_i=y_i-\exp(f_{m-1}(x_i)).
$$

Next, fit a regression tree to the inputs $x_i$ and these pseudo-responses $r_i$, using squared error:

$$
\boxed{h_m=\arg\min_{h\in\mathcal H}\sum_{i=1}^{n}(r_i-h(x_i))^2,}
$$

where $\mathcal H$ is the chosen class of regression trees.

Then choose a step size $\lambda_m$, for example by line search to minimize the Poisson loss:

$$
\lambda_m=\arg\min_{\lambda}\sum_{i=1}^{n}L\left(y_i,f_{m-1}(x_i)+\lambda h_m(x_i)\right).
$$

Update the predictor:

$$
\boxed{f_m(x)=f_{m-1}(x)+\lambda_m h_m(x).}
$$

Repeat this procedure to add more trees. Each new tree approximates the negative gradient of the current model's loss.
